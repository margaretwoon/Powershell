```python
"""
This program read the source file and replace X for each column 
Author : Margaret Woon
Version : 0.001
Date : 2026/06/03
"""
import os
import shutil
from datetime import datetime
import pandas as pd
import csv

source_file_name = 'CSTDAT11489PropertyOutput_20260612141150.csv'
source_file = f"C:/Users/miw/OneDrive - Great Southern Bank/Documents/py3/DataProgram/Input/{source_file_name}"

target_directory = "S:/Data Transfer/databricks/bancsext/uat/in/"

field_names = ['property_address', 'CL_MUDID','CL_GNAFPID','CL_Property_Type','CL_Unit_No','CL_Street_No','CL_Street_Name','CL_Suburb','CL_State','CL_Postcode']
def copy_files(source_file, target_directory):
    print(f"source file :{source_file}")

    print(f"target directory :{target_directory}")
    
    df_source = pd.read_csv(source_file)

    for field in field_names:
        df_source[field] = df_source[field].astype(str).str.replace('.', 'X', regex=True)

    df_source.to_csv(f"{target_directory}{source_file_name}", index = False, quoting=csv.QUOTE_ALL)

    return 

datetime_now = datetime.now().strftime("%d/%m/%Y, %H:%M:%S")
                        
print(f"Job Started at {datetime_now}\n")

# Initialize the output file
outputfile = f'TF0220_copied_files.txt'

if __name__ == "__main__":   

    csv_files_copied = copy_files(source_file, target_directory)


        
datetime_now = datetime.now().strftime("%d/%m/%Y, %H:%M:%S")
                        
print(f"Job Completed at {datetime_now}\n")
```
