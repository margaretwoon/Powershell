```python
import os
from docx import Document

import win32com.client

filelist = [
"ALIM",
"AUDT",
"BDEH",
"BIVM",
"BLBP",
"BLDVN1",
"BLDVPP",
"BLEP",
"BLIP",
"BOAF",
"BORG",
"BORZ",
"BPAY",
"BPBD",
"BSBM",
"CASM",
"CAST",
"CFDP",
"CFLG",
"CHPD",
"CHQM",
"CMPN",
"CSEC",
"CUEH",
"CUIH",
"CUSE",
"CUSVZZ",
"DDCM",
"DFRM",
"DISP",
"ED3P",
"EPCC",
"EXFM",
"FEEC",
"GECT",
"GLCP",
"GLDM",
"HOLD",
"HOLM",
"IADVGG",
"IMG0",
"INSM",
"INVV",
"ISCU",
"MILD",
"NPIM",
"NPOM",
"RBPM",
"REWD",
"RFND",
"SDDT",
"STAT",
"SUPP",
"SWRT",
"TDBO",
"TINM",
"UNCV",
"VAUT",
"VULN",
"YTDH",

]

def search_string_in_word_documents(folder_path, search_string):
    """
    Searches for a specific string in all Word documents within a folder.

    Args:
        folder_path (str): Path to the folder containing Word documents.
        search_string (str): The string to search for.

    Returns:
        list: List of file paths where the search string was found.
    """
    found_files = []

    # Iterate through all files in the folder
    for filename in os.listdir(folder_path):
        #if filename.endswith(".docx") or filename.endswith(".doc"):
        if filename.endswith(".doc"):
            file_path = os.path.join(folder_path, filename)
            try:
                doc = Document(file_path)
                for para in doc.paragraphs:
                    if search_string in para.text:
                        found_files.append(file_path)
                        break
            except Exception as e:
                print(f"Error reading {file_path}: {e}")

    return found_files


#folder_path = "/path/to/your/folder"



folder_path ='G:/mis/Application Delivery/Nucleus 1.0\Data Conversion/TCS Bancs File and Field Definitions/'
for index, file in enumerate(filelist):
    print(f"Processing {index} : {file}")
    search_string = file
    result = search_string_in_word_documents(folder_path, search_string)

    if result:
        print(f"Found in the following files:")
        for file_path in result:
            print(file_path)
    else:
        print(f"No occurrences of '{search_string}' found in any Word documents.")
```
