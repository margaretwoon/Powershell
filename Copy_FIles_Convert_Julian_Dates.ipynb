```python
"""
This program search through the entire source directory in network drive and sub directory and decode the julian day in the file name and create the date folder into rename the files before copy to a target network directory
Author : Margaret Woon
Version : 0.001
Date : 2026/06/04
"""
import os
import shutil
from datetime import datetime
import re

def add_dot_before_last_T(file_name):
    idx = file_name.rfind("T")
    if idx != -1:
        return file_name[:idx] + "." + file_name[idx:]
    return file_name

def scan_directory(from_directory, to_directory):
    copied_files = []
    print(f"source_directory = {from_directory}")
    print(f"target_directory = {to_directory}")
    for root, dirs, files in os.walk(from_directory):  
  
        #if os.path.basename(root).startswith(f"{date_now}"):
        print(f"dir_name : {os.path.basename(root)}")
        for file in files:
            #print(f"file name  {file}")
            if file.startswith('BNCDP01.TX0150.REPORT.D2025'):
                file_name_str = str(file)
                julian_str = re.search(r"D(\d{4})(\d{3})", file_name_str) # Day 162 of the year 2026
                print(f"julian_str {julian_str}")
                if julian_str:
                    year = julian_str.group(1)
                    julian_day = julian_str.group(2)
                    if year == '2025':
                        if int(julian_day) >= 121 and int(julian_day) <= 309:
                            print(f"file name = {file}")
                            print(f"year = {year}")
                            print(f"julian_day = {julian_day}")
                            actual_date1 = datetime.strptime(f"{year}{julian_day}", "%Y%j").date()
                            print(f"Actual Date: {actual_date1.strftime("%Y%m%d")}")  # Output: 20260611
                            target_folder_name = actual_date1.strftime("%Y%m%d")
                            target_folder = os.path.join(to_directory, target_folder_name)
                            try:
                                os.makedirs(target_folder, exist_ok = True)
                            except FileExistsError:
                                print(f"Error: A file already exists at {target_folder}")
                            except PermissionError:
                                print(f"Error: Insufficient permissions to create {target_folder}")
                            except OSError as e:
                                print(f"OS Error occurred: {e}")
                            finally:
                                target_path = os.path.join(target_folder, file)
                                #print(f"target_path {target_path}")
                                try:
                                    source_path = os.path.join(root, file)
                                    shutil.copy2(source_path, target_path)
                                except shutil.SameFileError:
                                    print("Source and destination are the exact same file.")
                                except FileNotFoundError:
                                    print("The source file could not be found.")
                                except PermissionError:
                                    print("Permission denied. Check file locks or folder write-rights.")
                                except OSError as e:
                                    print(f"Filesystem or metadata error occurred: {e}")
                                finally:
                                    copied_files.append(os.path.join(root, file)) 

    return copied_files


date_now = datetime.now().strftime("%Y%m%d")
print(f"Today's date : {date_now}")

datetime_now = datetime.now().strftime("%d/%m/%Y, %H:%M:%S")
                        
print(f"Job Started at {datetime_now}\n")
      
source_directory = r'\\sasnasprd001\PRD_Warehouse\01 Source\MIS_Files\DirectEntries'

#r'C:/Users/miw/OneDrive - Great Southern Bank/Documents/py3/Data/'

target_directory = r'S:/Data Transfer/databricks/bancsext/direct_entries/'

# Initialize the output file

outputfile = f'direct_entries_files_copied_output.txt'



if __name__ == "__main__":   

    matching_files = scan_directory(source_directory, target_directory)

    if matching_files:
        with open(outputfile,'w', encoding='utf-8') as output_file:
            for file_path in matching_files:
                output_file.write(f"{file_path}:\n")
                # output_file.write(f"\n\n")
        # Print a success message
        print(f"Copied Files Outcomes : {outputfile}\n")
    else:
        print(f"No files found with the string\n")

        
datetime_now = datetime.now().strftime("%d/%m/%Y, %H:%M:%S")
                        
print(f"Job Completed at {datetime_now}\n")
```
