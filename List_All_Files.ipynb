```python
"""
This program search through the entire source directory in network drive and sub directory and rename the files before copy to a target network directory
Author : Margaret Woon
Version : 0.001
Date : 2026/06/04
"""
import os
import shutil
from datetime import datetime

def add_dot_before_last_T(file_name):
    idx = file_name.rfind("T")
    if idx != -1:
        return file_name[:idx] + "." + file_name[idx:]
    return file_name

def scan_directory(from_directory):
    files_list = []
    print(f"source_directory = {from_directory}")

    for root, dirs, files in os.walk(from_directory):  
        print(f"dir_name : {os.path.basename(root)}")
        for file in files:
            if file.lower().endswith('.json'):
                #files_list.append(os.path.join(root, file)) 
                files_list.append(file)

    return files_list


date_now = datetime.now().strftime("%Y%m%d")
print(f"Today's date : {date_now}")

datetime_now = datetime.now().strftime("%d/%m/%Y, %H:%M:%S")
                        
print(f"Job Started at {datetime_now}\n")
      
source_directory = r'C:/Users/miw/OneDrive - Great Southern Bank/Documents/py3/DataProgram/Input'
#r'C:/Users/miw/OneDrive - Great Southern Bank/Documents/py3/Data/'



# Initialize the output file

outputfile = f'file_list_output.txt'



if __name__ == "__main__":   

    matching_files = scan_directory(source_directory)

    if matching_files:
        with open(outputfile,'w', encoding='utf-8') as output_file:
            for file_path in matching_files:
                output_file.write(f"{file_path}\n")
                #output_file.write(f"\n\n")
        # Print a success message
        print(f"File List Outcomes : {outputfile}\n")
    else:
        print(f"No files found with the string\n")

        
datetime_now = datetime.now().strftime("%d/%m/%Y, %H:%M:%S")
                        
print(f"Job Completed at {datetime_now}\n")
```
