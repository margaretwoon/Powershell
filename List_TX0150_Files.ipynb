```python
"""
This program search through the entire source directory in a given network drive and list all files contains *.TX0150.*.TXT between 20250501 and 20251107
Author : Margaret Woon
Version : 0.001
Date : 2026/06/19
"""
import os
import shutil
from datetime import datetime
import re

def add_dot_before_last_T(file_name):
    idx = file_name.rfind("T")
    if idx != -1:
        return file_name[:idx] + "." + file_name[idx:]
    return file_name

def scan_directory(from_directory, to_directory):
    files_list = []
    print(f"source_directory = {from_directory}")
    print(f"target_directory = {to_directory}")
    for root, dirs, files in os.walk(from_directory):  
        # print(f"root = {root}")
        # print(f"folder = {os.path.basename(root)}")
        # print(f"len(files) = {len(files)}")
        #if os.path.basename(root).startswith(f"{date_now}"):
        for file in files:
            file_name_str = str(file)
            if re.search(r"\.TX0150\.", file_name_str, re.IGNORECASE):
                #print(f"{file_name_str}")
                date_prefix = file_name_str[:6]
                #print(f"{date_prefix}")
                if "250501" <= date_prefix <= "251107":
                    find_str = re.search(r'(BNCDP01.*?\.TXT)',file_name_str, re.IGNORECASE)
                    if find_str:
                        trimmed_file_name = find_str.group(1)
                        with open(os.path.join(root, file), "r") as f:
                            record_cnt = sum(1 for line in f)      
                        files_list.append(f"{root}|{trimmed_file_name}|{record_cnt}")

    return files_list


date_now = datetime.now().strftime("%Y%m%d")
print(f"Today's date : {date_now}")

datetime_now = datetime.now().strftime("%d/%m/%Y, %H:%M:%S")
                        
print(f"Job Started at {datetime_now}\n")
      
source_directory = r'S:/Data Transfer/BaNCS/PROD/Business Drop Zone/Archive/'

#r'C:/Users/miw/OneDrive - Great Southern Bank/Documents/py3/Data/'

target_directory = r'S:/Data Transfer/databricks/bancsext/uat/in/'

# Initialize the output file

outputfile = f'TF0150_file_list_output.txt'



if __name__ == "__main__":   

    matching_files = scan_directory(source_directory, target_directory)

    if matching_files:
        with open(outputfile,'w', encoding='utf-8') as output_file:
            for file_path in matching_files:
                output_file.write(f"{file_path}\n")
                #output_file.write(f"\n\n")
        # Print a success message
        print(f"Files List Outcomes : {outputfile}\n")
    else:
        print(f"No files found with the string\n")

        
datetime_now = datetime.now().strftime("%d/%m/%Y, %H:%M:%S")
                        
print(f"Job Completed at {datetime_now}\n")
```
