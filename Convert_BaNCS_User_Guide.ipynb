```python
"""
   Step 1: Read the input USER GUIDE input file CUSC
   
"""
import pandas as pd

import numpy as np

import itertools 

import os

from datetime import datetime

i_fileName ='CUSC_UG.csv'

tableName =i_fileName.split("_")[0]

cwd = os.getcwd()

print(cwd)

input_directory = cwd+"\\Input\\"



#df = pd.read_csv(input_directory+i_fileName, skiprows=5, encoding='utf8')

try :
    df = pd.read_csv(input_directory+i_fileName, sep ='|',skiprows=5, header=0, encoding='utf8')
except :
    df = pd.read_csv(input_directory+i_fileName, sep ='|',skiprows=5, header=0, encoding='ISO-8859-1')
else :
    print("Error reading csv file")
    


working_directory = cwd + "\\" +"sql"+tableName.replace('[','').replace(']','')

# if not os.path.exists(working_directory):
#     os.makedirs(working_directory)
```

```python
"""
   Step 2: Output to CSV file.
   
"""
first_time = True

f=open("part01_cleaned"+tableName+".csv","w")

f.write("BUSINESS_DESCRIPTIONS"+"|"+
        "DB_ATTRIBUTES"+"|"+
        "UG_DESCRIPTIONS"+"|"+
        "\n")

now = datetime.now()

print("Job Started at ",now.strftime("%d/%m/%Y, %H:%M:%S"))
                       
descriptions_str=''


for row in df.itertuples():
    
    if first_time:
       
        print("row ", row)
        first_time = False
        if not pd.isna(row[1]):
            svrow = row
            print("Is Not NA")
        ### saved the row data away as reference
      
            print(row[1],",", row[2])
       
    else:
       
        if not pd.isna(row[1]):
            temp_str =  str(svrow[1])+"|"+\
                            str(svrow[2])+"|"+\
                            str(descriptions_str)+"|"\
                            +"\n"
            temp_str2 = temp_str.replace('nan','').replace('?','-') 
            descriptions_str=''
            #print("temp_str ", temp_str2)
            
            f.write(temp_str2)
            svrow = row
        else:
            descriptions_str = descriptions_str+' '+str(row[2])
                                
                            
temp_str =  str(svrow[1])+"|"+\
            str(svrow[2])+"|"+\
            str(descriptions_str)+"|"\
            +"\n"
temp_str2 = temp_str.replace('nan','').replace('?','-') 
f.write(temp_str2)
f.close() 

now = datetime.now()

print("Job Completed at ",now.strftime("%d/%m/%Y, %H:%M:%S"))
                       
    
        
```
