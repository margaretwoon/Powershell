```python
#from datetime import datetime, date
import os
import pandas as pd

dir_to_list = "C:\\Users\\miw\\Desktop\\py3\\DataProgram\\Bancs_Reference\\"



ctr_excel = 0
ctr_sheet = 0
cnt = 0

def process_file(input_f):

    ctr_sheet = 0
    xls = pd.ExcelFile(input_f)

    # PROCESS all TABS/SHEETS if ANY
    for sheetname in xls.sheet_names:
        ctr_sheet = ctr_sheet + 1
        print("**** ",'Sheet name: '+sheetname)
       
    print("Number of sheets/tabs processed: ", ctr_sheet,"\n")



print("\nSTARTING_DIR: ", dir_to_list,"\n")

for root, dirs, files in os.walk(dir_to_list, topdown=True):
    
    print("\nroot: ",root)
    print("\ndirs: ",dirs)
    print("\nfiles: ", files)
    
    writer = pd.ExcelWriter('out.xlsx', engine='xlsxwriter')
    
    for name in files:
        currentdir_str=os.path.join(root, name)
      #  if search_str.lower() in currentdir_str.lower() or search_str1.lower() in currentdir_str.lower() : 
       
        print("\nCurrent Dir: ",os.path.join(root, name))
        cnt += 1
        
        df = pd.read_csv(currentdir_str)
        df.to_excel(writer, sheet_name=os.path.splitext(os.path.basename(currentdir_str))[0], index=False)
        
        
    writer.save()
    print("cnt ", cnt)
       
        
            

print("\nDone processing!: ", ctr_excel)
```
