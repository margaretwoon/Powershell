```python
import requests
import json

# Configuration
ARCHER_URL = "https://gsb-test.archerirm.com.au" # Replace with your URL
INSTANCE_NAME = "710144"
USERNAME = "databricks_service_test"
PASSWORD = "Archer@123!!"

#https://gsb-test.archerirm.com.au/platformapi/core/security/login


# 1. Authenticate and get Session Token
login_url = f"{ARCHER_URL}/platformapi/core/security/login"
login_payload = {
   "InstanceName": INSTANCE_NAME,
   "Username": USERNAME,
   "Password": PASSWORD
}
headers = {"Content-Type": "application/json", "Accept": "application/json"}

print("Authenticating with Archer...")
response = requests.post(login_url, json=login_payload, headers=headers)
response.raise_for_status()
session_token = response.json()["RequestedObject"]["SessionToken"]

# 2. Fetch all Level IDs
level_url = f"{ARCHER_URL}/platformpi/core/system/level"
api_headers = {
   "Authorization": f"Archer session_token={session_token}",
   "Accept": "application/json"
}

print("Fetching Level IDs...")
level_response = requests.get(level_url, headers=api_headers)
level_response.raise_for_status()
levels = level_response.json()

# 3. Print out the dictionary map
print("\n--- Archer Level ID Dictionary ---")
for level in levels.get("RequestedObject", []):
   level_id = level.get("Id")
   level_name = level.get("Name")
   module_id = level.get("ModuleId") # This links it to the parent Application

   print(f"Application/Level Name: {level_name} | LevelID: {level_id} | ModuleID: {module_id}")

   # Highlight if it matches the ID you were worried about
   if level_id == 30465:
       print(f" >>> FOUND MATCH FOR ID 30465: Assigned to '{level_name}'")
```
