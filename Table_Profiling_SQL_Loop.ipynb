```python
"""
 SQL query for CUSM with OOP
 
"""


import pyodbc

import sqlalchemy

from datetime import datetime

import os
import pandas as pd

import sys

if r'C:\Users\miw\Desktop\py3\Utils' not in sys.path:
    sys.path.append(r'C:\Users\miw\Desktop\py3\Utils')

from DataProgramUtil import ServerName

#instantiate an object
my_server = ServerName()

my_cursor = my_server.sql_connect()

file='CUSM_FIELD_LIST.txt'

#open CUSM_FIELD_LIST.txt for reading
f=open(file,"r")

tableName = '[CUSM]'

# get the current working directory
cwd = os.getcwd()

working_directory = cwd + "\\" +"sql"+tableName.replace('[','').replace(']','')

if not os.path.exists(working_directory):
    os.makedirs(working_directory)

field_list = f.readlines()

#print(field_list)

clean_list =[]

cnt = 0



for line in field_list:
    clean_list.append(line.replace('\n', ''))
    
for j in clean_list: 
    
    cnt += 1
    
    print("Job started for ", j ," at ",datetime.now().strftime("%d/%m/%Y, %H:%M:%S"))
    print("\n")
        
    sql_groupby ='SELECT '+j+', count(*) as the_count FROM [EDH_PSA_PRD].[BANCS].'+tableName+\
    ' where [FlexRowIsCurrent]=1 group by '+j+' order by '+j
    print("sql_groupby : ", sql_groupby)
    
    
    try:
        df = pd.read_sql(sql_groupby, my_cursor)
    except:
        print("sql failed at ", sql_groupby)
    else:
        csv_output_filename = working_directory+"\\"+j.replace('[','').replace(']','')+'_groupby.csv'
        df.to_csv(csv_output_filename,sep="|", index=False)
    

    print("Job Completed for ", j ," at ",datetime.now().strftime("%d/%m/%Y, %H:%M:%S"))
    print("\n")
#     if cnt >  10:
#         break

print("\n","*"*100)
print("ALL query Completed for ", j ," at ",datetime.now().strftime("%d/%m/%Y, %H:%M:%S"))
```
