```python
"""
This program search through the targeted Network drive and list the directory and subdirectory
Author : Margaret Woon
Version : 0.001
Date : 2026/06/03
"""
import os
import shutil
from datetime import datetime

source_directory = "S:/Automation_CFO/3.OPERATIONAL_RECONCILIATION/01.REPORTS/01.BaNCS_REPORTS/05.TF0220/"

target_directory = "S:/Data Transfer/databricks/bancsext/uat/in/"

def add_dot_before_last_T(file_name):
    idx = file_name.rfind("T")
    if idx != -1:
        return file_name[:idx] + "." + file_name[idx:]
    return file_name

def copy_files(source_directory, target_directory):
    print(f"source directory :{source_directory}")

    print(f"target directory :{target_directory}")
    copied_files = []
    os.makedirs(target_directory, exist_ok=True)
    for directory in source_directory:
        for root, dirs, files in os.walk(directory):
            for file in files:
                print(f'file {file}')
                if file.lower().endswith('.csv'):
                    if os.path.basename(root).startswith("20260603"):
                        source_path = os.path.join(root, file)
                        print(f"source_path {source_path}")
                        copied_file_name = add_dot_before_last_T(file)
                        target_path = os.path.join(target_directory, copied_file_name)
                        print(f"target_path {target_path}")
                        shutil.copy2(source_path, target_path)
                        print(f"root {root}, file {file}")
                        copied_files.append(os.path.join(root, file))  
    return copied_files

datetime_now = datetime.now().strftime("%d/%m/%Y, %H:%M:%S")
                        
print(f"Job Started at {datetime_now}\n")

# Initialize the output file
outputfile = f'TF0220_copied_files.txt'

if __name__ == "__main__":   

    csv_files_copied = copy_files(source_directory, target_directory)

    if csv_files_copied:
        with open(outputfile,'w', encoding='utf-8') as output_file:
            for file_path in csv_files_copied:
                output_file.write(f"{file_path}:\n")
                output_file.write(f"\n")
        # Print a success message
        print(f"Outcomes : {outputfile}\n")
    else:
        print(f"No files found in {source_directory}\n")

        
datetime_now = datetime.now().strftime("%d/%m/%Y, %H:%M:%S")
                        
print(f"Job Completed at {datetime_now}\n")
```
