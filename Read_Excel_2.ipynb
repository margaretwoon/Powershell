```python
"""
This program read 
Author : Margaret Woon
Version : 0.001
Date : 2026/8/26
"""
import os
import shutil
from datetime import datetime
import pandas as pd
from pathlib import Path
import time
import csv


# Load the Excel file
file_path = r"file:///"C:\Users\miw\OneDrive - Great Southern Bank\Documents\py3\DataProgram\Input\CANCELLED_LEAVE_TABLE.CSV""

# Read the excel file
xls = pd.ExcelFile(file_path)

target_tab_selection = ['CANCELLED_LEAVE_TABLE']

for sheet in xls.sheet_names:
        df = pd.read_excel(file_path, sheet_name = sheet)

        print(f"Tab : {sheet}")

        if sheet in target_tab_selection:
            print(f"df {df.columns}")

            # remove spaces from column names
            df.columns = [col.strip() if isinstance(col, str) else col for col in df.columns]

            
            if sheet == target_tab_selection[0]:
                #output_df_cols = ['Business Area','Team', 'Measure', 'Target'] 
                output_df_cols = [df.columns[1], df.columns[2], df.columns[3], df.columns[4]]
            elif sheet == target_tab_selection[1]:
                #output_df_cols = ['Business Area','Team', 'Measure', 'Target', 'Effective From Date']
                output_df_cols = [df.columns[1], df.columns[2], df.columns[3], df.columns[5], df.columns[8]]
                    

            print(f"output_df_cols : {output_df_cols}")
            # identify all the date columns
            date_cols = [col for col in df.columns if isinstance(col, (pd.Timestamp, datetime))]

            # convert blank cells to NA
            df[output_df_cols] = (df[output_df_cols].replace(r'____', pd.NA, regex=True))
        
            # remove rows where all first 5 columns are blank
            df_cleaned = df.dropna(
                subset = output_df_cols,
                        how='all')

            df_long = df_cleaned.melt(
                id_vars = output_df_cols,
                value_vars = date_cols,
                var_name = 'Date',
                value_name='Metrics'
            )

            print(f"df_long :{df_long}")

            output_path = f"./output/{sheet}.csv"

            df_long.to_csv(output_path, sep=',' ,index=None, quoting=csv.QUOTE_ALL)


```
